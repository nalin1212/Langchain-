# Streaming and Batch 


##  Streaming 

Most models can stream thier output content while it is being generated .By displaying output Progressively ,streaming significantly 
improves user experience ,particularly for longer responses.Calling stream () returns an iterator that yeilds output chunks as they are 
produced. you can use a loop to process each chunk in real time .

```python
model.invoke("write me a 200 paragraph on AI")
```

```python
model.stream("write me a 200 paragraph on AI")
```

```python
for chunk in model.stream("write me a 50 paragraph on AI"):
    print(chunk.content, end=" ",flush=True)
```

```python
for chunk in model.stream("why do parrots talk"):
    print(chunk.text, end="")
```

## Batch 
### collection of independent requests to a model can significantly improve performance and reduce costs , as the processing can be done in parallel


```python
response =model.batch(["why do parrots talk","why do dogs bark,what is quantum computing?"])
for response in response:
    print(response.content)
```

```python
model.batch (["why do parrots talk","why do dogs bark,what is quantum computing?"],
             config={'max_concurrency':5, # limit to 5 parrlel calls 
                     })
```
